# Story Generation Model Evaluation Report

## 1. Objective

The purpose of this evaluation is to compare the storytelling performance of multiple Large Language Models (LLMs) under different prompt versions. The evaluation measures both generation quality and inference latency.

Models evaluated:

- Gemini 3 Flash
- Llama 3.3 70B Versatile

Prompt variants evaluated:

- Prompt Version 1 (v1)
- Prompt Version 2 (v2)

Each model–prompt combination was tested on the same set of story-generation inputs and evaluated using an LLM-as-a-Judge approach.

---

```python
import pandas as pd
```

## 2. Evaluation Methodology

### 2.1 Dataset

The evaluation dataset consists of generated children's stories produced from a fixed set of story prompts.

For each generated story, the following information was recorded:

- Model used
- Prompt version
- Generation latency
- Story structure validation
- Quality evaluation scores

---

### 2.2 Quality Metrics

Stories were scored on a 1–10 scale across five dimensions:

| Metric | Description |
|----------|-------------|
| Coherence | Logical flow and readability of the story |
| Consistency | Narrative consistency and absence of contradictions |
| Creativity | Originality and imagination |
| Progression | Story development and pacing |
| Language | Grammar, fluency, and language quality |

---


```python
df = pd.read_csv("checkpoint3.csv")
df
```

```python
df =df.drop("run_number", axis=1)
df
```

### 2.3 Composite Score

A weighted overall score was calculated:

\[
TotalScore =
0.30 (Coherence) 
+0.25 (Consistency) 
+0.20 (Creativity) 
+0.15 (Progression) 
+0.10 (Language) 
\]

### Why a Weighted Score Was Used

Story quality is a multidimensional concept that cannot be accurately represented by a single evaluation criterion. While metrics such as coherence, creativity, consistency, progression, and language quality are all important, they do not contribute equally to the overall quality of a children's story.

A weighted scoring approach was therefore adopted to reflect the relative importance of each storytelling dimension:

- **Coherence (30%)** received the highest weight because a story must remain understandable and logically connected for readers to follow the narrative.
  
- **Consistency (25%)** was heavily weighted since maintaining character behavior, facts, and narrative details throughout the story is essential for story quality.

- **Creativity (20%)** was assigned a substantial weight because engaging and imaginative content is a key characteristic of children's storytelling.

- **Progression (15%)** measures how effectively the story develops and moves toward a conclusion.

- **Language Quality (10%)** was given the lowest weight because all evaluated models generally produced grammatically correct text, making it a less discriminative factor than narrative quality.

The weighted score provides a single aggregate measure that captures overall storytelling performance while prioritizing the aspects that have the greatest impact on reader experience. This enables a more meaningful comparison between models and prompt variants than treating all evaluation dimensions as equally important.   

The weighting scheme was designed based on the objectives of children's story generation, where narrative coherence and consistency are considered more critical than surface-level language quality.

---

```python
df['total_score'] = ( df["coherence"] * 0.30 +
                      df["consistency"] * 0.25 +
                      df["creativity"] * 0.20 +
                      df["progression"] * 0.15 +
                      df["language"] * 0.10
                    )
df.head(10)
```


### Additionally, the following metrics were calculated:

- **Best Input Score**: Highest total score achieved on any story input.
- **Worst Input Score**: Lowest total score achieved on any story input.
- **Score Volatility**: Standard deviation of total scores across inputs.

Lower volatility indicates more consistent performance.

```python
df_piv = df.groupby(["model", "prompt_version"]).agg(
    latency=('latency', 'mean'),
    coherence=('coherence', 'mean'),
    consistency=('consistency', 'mean'),
    creativity=('creativity', 'mean'),
    progression=('progression', 'mean'),
    language=('language', 'mean'),
    total_score=('total_score', 'mean'),
    
    best_input_score=('total_score', 'max'),
    worst_input_score=('total_score', 'min'),
    score_volatility=('total_score', 'std')
).reset_index()

df_piv
```

# 10. Deployment and Rate Limit Considerations

In addition to quality and latency, API usage limits were considered when comparing the evaluated models.

| Model | Requests / Minute | Requests / Day |
|---------|---------|---------|
| Gemini 3 Flash | 5 RPM | 20 RPD |
| Llama 3.3 70B Versatile | 30 RPM | 1,000 RPD |

These limits have a direct impact on the scalability of a story generation system. While Gemini 3 Flash achieved the highest storytelling quality, its significantly lower request quotas restrict the number of stories that can be generated within a given time period. This can become a bottleneck for applications serving multiple users or requiring large-scale content generation.

In contrast, Llama 3.3 70B Versatile offers substantially higher throughput, supporting six times more requests per minute and fifty times more requests per day. Although its story quality scores were lower, its higher rate limits make it more suitable for high-volume deployment scenarios.
