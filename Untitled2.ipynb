```python
import pandas as pd
```

```python
df = pd.read_csv("checkpoint3.csv")
df
```

```python
df =df.drop("run_number", axis=1)
df
```

```python
df['total_score'] = ( df["coherence"] * 0.30 +
                      df["consistency"] * 0.25 +
                      df["creativity"] * 0.20 +
                      df["progression"] * 0.15 +
                      df["language"] * 0.10
                    )
df.head(10)
```

```python
avg_latency_per_model = df.groupby(['model','prompt_version'])['latency'].mean()
avg_latency_per_model
```

```python
avg_coherence_per_model = df.groupby(['model','prompt_version'])['coherence'].mean()
avg_coherence_per_model
```

```python
avg_consistency_per_model = df.groupby(['model','prompt_version'])['consistency'].mean()
avg_consistency_per_model
```

```python
avg_creativity_per_model = df.groupby(['model','prompt_version'])['creativity'].mean()
avg_creativity_per_model
```

```python
avg_progression_per_model = df.groupby(['model','prompt_version'])['progression'].mean()
avg_progression_per_model
```

```python
avg_language_per_model = df.groupby(['model','prompt_version'])['language'].mean()
avg_language_per_model
```

```python
avg_total_score_per_model = df.groupby(['model','prompt_version'])['total_score'].mean()
avg_total_score_per_model
```

```python
cols = ["latency", "coherence","consistency", "creativity", "progression", "language", "total_score"]
df_pivot = df.groupby(["model", "prompt_version","story_input"])[cols].mean().reset_index()
df_pivot
```

```python
df_piv = df.groupby(["model", "prompt_version"]).agg(
    latency=('latency', 'mean'),
    coherence=('coherence', 'mean'),
    consistency=('consistency', 'mean'),
    creativity=('creativity', 'mean'),
    progression=('progression', 'mean'),
    language=('language', 'mean'),
    total_score=('total_score', 'mean'),
    
    best_input_score=('total_score', 'max'),
    worst_input_score=('total_score', 'min'),
    score_volatility=('total_score', 'std')
).reset_index()

df_piv
```
